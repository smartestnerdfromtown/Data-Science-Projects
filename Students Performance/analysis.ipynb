```python
import pandas as pd
import numpy as np
import os

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
os.makedirs("figures", exist_ok=True)
```

```python
df = pd.read_csv(
    filepath_or_buffer="student_performance_cleaned.csv"
)
```

```python
df.info()
```

```python
title = "Distribution of Student Performance Index"

plt.figure(figsize=(8, 5))

sns.histplot(
    df["performance_index"],
    bins=30,
    kde=True,
    color="#4C72B0",
    edgecolor="white",
    linewidth=0.5
)

plt.xlabel("Performance Index")
plt.ylabel("Number of Students")
plt.title(title, fontsize=15, weight="bold", pad=15)

plt.tight_layout()

filename = "figures/performance_index_distribution.png"
plt.savefig(filename, dpi=300, bbox_inches="tight")
plt.show()
```

```python
df["extracurricular_activities"].value_counts()
```

```python
activity_counts = df["extracurricular_activities"].value_counts()

title = "Extracurricular Activity Participation"

plt.figure(figsize=(6, 6))

plt.pie(
    activity_counts,
    labels=activity_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=["#4C72B0", "#55A868"]
)

plt.title(title, fontsize=14, weight="bold")
plt.tight_layout()

filename = "figures/extracurricular_pie.png"
plt.savefig(filename, dpi=300, bbox_inches="tight")
plt.show()
```
